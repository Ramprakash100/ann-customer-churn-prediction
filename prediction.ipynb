```python
import tensorflow as tf 
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
##Load the trained model, scale pickle, onehot encoders
model = load_model('model.h5', compile=False)

#Load the encoder and scale
with open('onehot_encoder_geo.pkl','rb') as file:
    lable_encoder_geo = pickle.load(file)

with open('label_encoder_gender.pkl','rb') as file:
    lable_encoder_gender = pickle.load(file)

with open('scale.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
##One hot encode geography
geo_encoded = lable_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded,columns =lable_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender'] = lable_encoder_gender.transform([input_data['Gender']])
input_df
```

```python
input_df.columns
```

```python
input_df
```

```python
input_scale = scaler.transform(input_df)
input_scale
```

```python
prediction = model.predict(input_scale)
prediction
```
